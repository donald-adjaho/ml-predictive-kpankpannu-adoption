# ETAPE 2: NETTOYAGE DE LA BASE DE DONNÉE
Nous allons maintenant nous servir du plan de nettoyage que notre agent IA nous a fourni pour notre base. 

Avant de passer au plan, j'aimerais présenter quelques questions très importantes que je devrais répondre pour permettre à l'agent de me fournir un code propre:
- C-IM1:
*Merci pour cette analyse. Cependant, dans notre contexte d'étude au Nord-Bénin, le 0 n'est pas une valeur manquante, c'est une réalité socio-économique : beaucoup de femmes sont au foyer et n'ont pas de revenu personnel, mais elles ont accès au revenu du ménage.*
Applique donc la stratégie suivante :
    - Garde les 0 tels quels dans la colonne income (ne les convertis pas en NaN).
    - Crée une variable indicatrice is_financially_dependent (1 si income == 0, sinon 0).
    - Assure-toi que la variable household_income (revenu du ménage) est bien présente et traitée numériquement. C'est cette variable qui sera la plus prédictive pour notre modèle ML.
- C-IM2:
  Je valide l'option 1. Pourquoi crée donc une colonne de traçabilité ?
  Au lieu de modifier ou supprimer silencieusement la valeur aberrante (15 enfants pour une femme de 27 ans), nous la plafonnons à un maximum plausible (5) tout en gardant une trace de l'intervention. Cela garantit l'intégrité des données, permet un retour en arrière facile si nécessaire, et peut même servir de variable informative pour le modèle de Machine Learning
- C-IM3:
  Je valide l'option 1 . en plus : Crée une colonne de traçabilité price_corrected (1 pour la ligne corrigée, 0 sinon).
- C-IM4:
  Excellente détection. Ces valeurs ne sont pas des erreurs de frappe, mais probablement un mélange de réponses stratégiques (négociation) ou d'une confusion avec le prix d'une petite mesure. Applique la stratégie suivante :
- Conserve les valeurs telles quelles dans ideal_price (on ne les modifie pas).
- Crée une variable indicatrice low_price_respondent (1 si ideal_price <= 500, sinon 0). Cela nous permettra de dire au modèle ML : "Attention, ces personnes ont un comportement d'achat très spécifique".
- Ne les exclus pas de la base maintenant, mais sache que dans le Notebook 03 (Machine Learning), nous exclurons ces lignes spécifiquement lors de l'entraînement du modèle de régression sur le prix, pour ne pas fausser la courbe.

Présentation du Plan de nettoyage 

Plan de nettoyage — ce que fait mon code, étape par étape
 1. Chargement : lit le fichier brut base_femmes_2024.xlsx (106 lignes, 138 colonnes). Assertion : forme exacte (106, 138).
 2. Nettoyage des noms de colonnes : enlève les espaces en début/fin de chaque nom de colonne (Kobo laisse souvent des espaces).
 3. Sélection des variables : ne garde que les 9 variables cibles demandées + household_income + _id pour la traçabilité. Toutes les autres colonnes sont supprimées.
 4. Renommage : remplace les noms français par des noms anglais (age, income, household_income, education, is_pregnant, has_child_under_5, num_children, buy_intent, ideal_price, perceived_child_benefit).
 5. Variables binaires : convertit is_pregnant, has_child_under_5 et buy_intent de Oui/Non en 1/0. Assertion : pas de valeur autre que Oui/Non, pas de NaN.
 6. Niveau d'instruction : education normalisé (accents, casse) et passé en catégorie (Aucun, Primaire, Secondaire, Supérieur).
 7. Revenus : income et household_income restent numériques, les 0 sont gardés tels quels (réalité socio-économique validée). Créé l'indicateur is_financially_dependent (1 si income == 0).
 8. Prix acceptable : ideal_price mis en numérique. L'erreur de frappe 15001 est corrigée en 1500 (division par 10) et signalée par price_corrected. Les 9 valeurs manquantes de `ideal_price` seront imputées par la médiane. La traçabilité est assurée par deux colonnes :
- `ideal_price_missing` (créée en amont) : identifie les non-réponses initiales.
- `ideal_price_imputed` (créée lors de l'imputation) : identifie explicitement les lignes dont la valeur a été imputée.
 9. Nombre d'enfants : num_children mis en numérique. La valeur aberrante 15 est plafonnée à 5 (num_children_outlier = 1). Les 29 NaN correspondent exactement aux femmes sans enfant de moins de 5 ans (logique de saut du questionnaire) — ils restent manquants et sont signalés par num_children_missing.
10. Indicateur prix bas : créé low_price_respondent (1 si ideal_price ≤ 500 FCFA, sinon 0). Ces lignes ne sont pas supprimées (exclusion réservée à la régression de prix dans le Notebook 03).
11. Vérification de la population : assertion que chaque femme est enceinte et/ou mère d'un enfant de moins de 5 ans.
12. Export : sauvegarde data/processed/base_femmes_2024_clean.csv (106 lignes × 18 colonnes).

### 🔬 Tests additionnels (Human-in-the-Loop)
Ces tests permettront de validé en profondeur la qualité de notre base de donnée après le nettoyage automatisé par IA
**EN:** Additional domain-driven tests validating data integrity beyond the automated suite.
**FR:** Tests additionnels guidés par l'expertise métier validant l'intégrité au-delà de la suite automatisée.

```python
def test_no_leakage_of_metadata():
    """EN: No KoboToolbox metadata columns (_uuid, _submission_time...) leaked.
    FR: Aucune colonne de métadonnées KoboToolbox ne doit avoir fuité."""
    forbidden = [c for c in df_clean.columns if c.startswith("_")]
    assert not forbidden, f"Metadata columns leaked: {forbidden}"


def test_household_income_coherence():
    """EN: Household income must be >= personal income for every respondent.
    FR: Le revenu du ménage doit être >= au revenu personnel pour chaque répondante."""
    incoherent = df_clean["household_income"] < df_clean["income"]
    assert not incoherent.any(), f"{incoherent.sum()} incohérences income/household_income"


def test_no_duplicate_respondents():
    """EN: No duplicate respondent_id in output.
    FR: Aucun respondent_id dupliqué en sortie."""
    assert df_clean["respondent_id"].is_unique


def test_education_categories_are_ordered():
    """EN: Education must only contain the 4 known levels.
    FR: L'éducation ne doit contenir que les 4 niveaux connus."""
    allowed = {"Aucun", "Primaire", "Secondaire", "Supérieur"}
    actual = set(df_clean["education"].dropna().astype(str).unique())
    assert actual <= allowed, f"Niveaux inconnus : {actual - allowed}"


def test_target_buy_intent_is_balanced():
    """EN: buy_intent should not be 100% one class (would make ML trivial).
    FR: buy_intent ne doit pas être 100% d'une seule classe (rendrait le ML trivial)."""
    counts = df_clean["buy_intent"].value_counts()
    assert len(counts) == 2, "buy_intent is a single class"
    ratio = counts.min() / counts.sum()
    assert ratio >= 0.05, f"Minority class only {ratio:.1%} — too imbalanced"


def test_no_infinite_values():
    """EN: No infinite values in numeric columns.
    FR: Aucune valeur infinie dans les colonnes numériques."""
    numeric = df_clean.select_dtypes(include="number")
    n_inf = (numeric == float("inf")).sum().sum() + (numeric == float("-inf")).sum().sum()
    assert n_inf == 0, f"{n_inf} valeurs infinies détectées"


# EN: Run each additional test and report
# FR: Exécute chaque test additionnel et rapporte le résultat
additional_tests = [
    test_no_leakage_of_metadata,
    test_household_income_coherence,
    test_no_duplicate_respondents,
    test_education_categories_are_ordered,
    test_target_buy_intent_is_balanced,
    test_no_infinite_values,
]

passed, failed = 0, 0
for t in additional_tests:
    try:
        t()
        print(f"✅ PASS : {t.__name__}")
        passed += 1
    except AssertionError as e:
        print(f"❌ FAIL : {t.__name__} → {e}")
        failed += 1

print(f"\n📊 Résultat : {passed} passés, {failed} échoués sur {len(additional_tests)}")
assert failed == 0, "Certains tests additionnels ont échoué."
```

### Commentaire des résultats
Tout les tests passent sauf le deuxième qui vérifie si le revenu personnel est inférieur au revenu du menage. Il y a une ligne de notre base qui ne respecte pas cette régle. Le revenu personnel ne peut pas dépasser celui du ménage , ce n'est mathématiquement pas possible. D'où l'importance de procéder à des validations via des tests stricts. Essayons d'investiguer sur cette incohérence dans notre base.

```python
### Identifions la ligne fautive qui fausse notre test
# EN: Identify the row where household_income < income
# FR: Identifie la ligne où le revenu du ménage est inférieur au revenu personnel
suspects = df_clean[df_clean["household_income"] < df_clean["income"]]
print(f"🔍 {len(suspects)} ligne(s) incohérente(s) :")
display(suspects[["respondent_id", "income", "household_income", "age", "education", "buy_intent"]])
```

La ligne 48 est donc la ligne responsable de l'échec du test. Corrigeons donc cette ligne.

```python
# EN: Detect ALL rows where household_income < income (logically impossible)
# FR: Détecte TOUTES les lignes où le revenu ménage < revenu personnel (logiquement impossible)
incoherent_mask = df_clean["household_income"] < df_clean["income"]
print(f"🔍 {incoherent_mask.sum()} ligne(s) incohérente(s) détectée(s)")
display(df_clean.loc[incoherent_mask, ["respondent_id", "income", "household_income"]])

# EN: Flag the correction for traceability
# FR: Marque la correction pour la traçabilité
df_clean["household_income_corrected"] = incoherent_mask.astype("Int64")

# EN: Household income must be at least equal to personal income (business rule)
# FR: Le revenu du ménage doit être au moins égal au revenu personnel (règle métier)
df_clean.loc[incoherent_mask, "household_income"] = df_clean.loc[incoherent_mask, "income"]

# EN: Verify the correction
# FR: Vérifie la correction
assert (df_clean["household_income"] >= df_clean["income"]).all(), "❌ Incohérence restante"
print(f"✅ {incoherent_mask.sum()} ligne(s) corrigée(s). Cohérence rétablie.")
```

### Commentaire du résultat
Super!!, La ligne 48 a été corrigér. Touts les test passent maintenant au vert. Cette étape de nettoyage et de transformation de nos données est maintenant cloturée.
